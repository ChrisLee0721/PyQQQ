# 教程 01：基础

学习 QuoNic 的基础：构建电路、运行电路、读取结果。

## 最简单的电路

```python
from quonic import qgate, qshow
from quonic.gates import H, CX

qgate(H, 0)      # 对量子比特 0 施加 Hadamard 门
qgate(CX, 0, 1)  # CNOT：控制=0，目标=1
qshow()           # 运行并显示结果

# 这创建了一个 Bell 态。qshow() 在最佳可用后端上运行电路并打印结果。
```

## 理解量子比特

```python
在 QuoNic 中，量子比特就是数字。qgate(H, 0) 对量子比特 0 施加 Hadamard 门。不需要创建电路对象，不需要管理量子比特寄存器。

qgate(X, 0)      # Pauli-X（比特翻转）作用于量子比特 0
qgate(H, 1)      # Hadamard 作用于量子比特 1
qgate(CX, 0, 1)  # CNOT：控制=0，目标=1
```

## 读取结果

```python
qshow() 返回一个 Result 对象，包含测量计数：

result = qshow()
print(result.counts)  # {'00': 512, '11': 512}
print(result.shots)   # 1024
```

## 切换后端

```python
同一个电路可以在不同后端上运行：

qshow(backend='qiskit')      # IBM Qiskit
qshow(backend='cirq')        # Google Cirq
qshow(backend='qulacs')      # Qulacs（高性能 C++）
qshow(backend='native')      # QuoNic 原生引擎
```

## 添加噪声

```python
模拟真实硬件的噪声：

qshow(noise=0.05)  # 5% 去极化噪声
```
