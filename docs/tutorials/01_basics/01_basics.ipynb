# Tutorial 01: QuoNic Basics

Learn the fundamentals: building circuits, running them, and reading results.

```python
from quonic import qgate, qshow
from quonic.gates import CX, H

qgate(H, 0)
qgate(CX, 0, 1)
result = qshow()
```

## GHZ State

```python
from quonic import reset

reset()
qgate(H, 0)
qgate(CX, 0, 1)
qgate(CX, 1, 2)
qshow()
```

## Reading Results

```python
reset()
qgate(H, 0)
qgate(CX, 0, 1)
result = qshow(shots=2048)
print(f'Counts: {result.counts}')
```

## Switching Backends

The same code runs on any backend.

```python
qshow(backend='native')
```
