# Tutorial 01: QuoNic Basics

Learn the fundamentals: building circuits, running them, and reading results.

## The Simplest Circuit

QuoNic makes quantum programming as simple as writing Python. No `QuantumCircuit` to learn, no `backend` to understand.

```python
from quonic import qgate, qshow
from quonic.gates import H, CX

# Bell state: H on qubit 0, then CNOT
qgate(H, 0)
qgate(CX, 0, 1)
result = qshow()  # Run and display
```

## Understanding Qubits

In QuoNic, qubits are just numbers. No qubit register to manage.

```python
from quonic import reset
from quonic.gates import X, Y, Z

# Pauli gates
reset()
qgate(X, 0)  # Bit flip
qgate(H, 1)  # Hadamard
qgate(CX, 0, 1)
qshow()
```

## GHZ State

A 3-qubit GHZ state: |000> + |111>.

```python
reset()
qgate(H, 0)
qgate(CX, 0, 1)
qgate(CX, 1, 2)
qshow()  # |000> and |111> each ~50%
```

## Reading Results

`qshow()` returns a `Result` object with measurement counts.

```python
reset()
qgate(H, 0)
qgate(CX, 0, 1)
result = qshow(shots=2048)
print(f"Counts: {result.counts}")
print(f"Shots: {result.shots}")
```
