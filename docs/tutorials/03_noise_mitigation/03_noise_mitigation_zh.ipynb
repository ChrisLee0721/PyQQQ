# 教程 03：噪声缓解

真实量子硬件有噪声。QuoNic 提供两种错误缓解技术。

## 添加噪声

```python
from quonic import qgate, qshow, reset
from quonic.gates import CX, H

reset()
qgate(H, 0)
qgate(CX, 0, 1)

# 无噪声：理想 Bell 态
print('无噪声：')
qshow()

# 5% 去极化噪声
print('有噪声：')
qshow(noise=0.05)
```

## ZNE 零噪声外推

```python
from quonic import zne
from quonic.ir import Circuit, GateOperation

c = Circuit()
c.add(GateOperation('x', (0,)))
c.add(GateOperation('measure', (0,)))

# ZNE 线性外推
res = zne(c, noise=0.05, target='1', shots=4096, extrapolation='linear')
print(f'原始: {res.values[0]:.3f}')
print(f'ZNE: {res.extrapolated:.3f}')
print('理想: 1.000')
```
