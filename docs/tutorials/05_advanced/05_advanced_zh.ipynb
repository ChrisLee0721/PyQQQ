# 教程 05：高级功能

探索 QuoNic 的高级功能：qif、creg、cwhile、优化。

## 量子条件语句 qif

```python
from quonic import qif, qgate, reset
from quonic.gates import H, X

reset()
qgate(H, 0)  # 叠加态
qif(0).then(X, 1)  # 如果 qubit 0 是 |1>，翻转 qubit 1
qshow()
```

## 经典条件语句 cif

```python
from quonic import cif, creg, qgate, reset
from quonic.gates import H, X

reset()
qgate(H, 0)
flag = creg('flag')
flag.measure(0)
cif(flag, 1).then(X, 1)  # 如果测量结果是 1，翻转 qubit 1
qshow()
```

## 电路优化

```python
from quonic.compiler import optimize
from quonic.ir import Circuit, GateOperation

c = Circuit()
c.allocate(1)
c.add(GateOperation('h', (0,)))
c.add(GateOperation('h', (0,)))  # H·H = I

optimized = optimize(c, passes=('cancel',))
print(f'优化前: {len(c.ops)} 门')
print(f'优化后: {len(optimized.ops)} 门')
```

## 自定义门

```python
from quonic.gates import Gate
import numpy as np

# 创建自定义 T 门
t_mat = np.array([[1, 0], [0, np.exp(1j * np.pi / 4)]], dtype=complex)
Gate.from_matrix('my_t', t_mat)

# 使用自定义门
qgate('my_t', 0)
qshow()
```
