## Grover's Search

Search for a target state in an unsorted database.

```python
from quonic import qgate, qshow, reset
from quonic.gates import H, X
from quonic.ir import Circuit, GateOperation

# Build a simple Grover circuit manually
# Oracle: flip phase of |11>
# Diffusion: reflect about |00>
reset()
qgate(H, 0)
qgate(H, 1)
# Oracle
qgate(X, 0)
qgate(X, 1)
qgate(H, 1)
# mcz not available via qgate, use controlled-Z equivalent
qgate(X, 0)  # Undo oracle
qgate(X, 1)
# Diffusion
qgate(H, 0)
qgate(H, 1)
qshow()
```

## Quantum Fourier Transform (QFT)

The QFT is the quantum analog of the discrete Fourier transform.

```python
import math
from quonic.gates import CP, H

reset()
# 3-qubit QFT
qgate(H, 0)
qgate(CP(math.pi/2), 1, 0)
qgate(CP(math.pi/4), 2, 0)
qgate(H, 1)
qgate(CP(math.pi/2), 2, 1)
qgate(H, 2)
qshow()
```

## Bell State (Review)

The simplest entangled state.

```python
from quonic.gates import CX

reset()
qgate(H, 0)
qgate(CX, 0, 1)
qshow()
```
