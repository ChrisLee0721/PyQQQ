# 教程 02：算法

探索 QuoNic 的 75 个算法模板。

## Grover 搜索

```python
from quonic.algorithms import grover

result = grover('11', 2, shots=1024)
print(result.counts)  # {'11': ~1000}

# Grover 搜索在无序数据库中找到目标，比经典方法快 √N 倍。
```

## 量子傅里叶变换

```python
from quonic.algorithms import qft

result = qft(n_qubits=3, shots=1024)
print(result.counts)

# QFT 是许多量子算法的基础，包括 Shor 算法和量子相位估计。
```

## VQE 变分量子本征求解器

```python
from quonic.algorithms import vqe

hamiltonian = [(1.0, 'ZZ'), (0.5, 'XI'), (0.5, 'IX')]
result = vqe(hamiltonian, n_qubits=2, maxiter=200)
print(f'基态能量: {result.value}')

# VQE 用于量子化学，计算分子的基态能量。
```

## QAOA 量子近似优化

```python
from quonic.algorithms import qaoa_maxcut

edges = [(0, 1), (1, 2), (0, 2)]
result = qaoa_maxcut(edges, 3, p=1, maxiter=200)
print(f'MaxCut: {result.value}')

# QAOA 用于组合优化问题，如 MaxCut、旅行商问题等。
```
