# QuoNic 教程

**量子编程，像写 Python 一样简单。**

本教程用可运行的 Jupyter 单元带你走一遍核心概念：
两个词（`qgate` 加门、`qshow` 看结果）就能跑通量子程序。

> 环境：`pip install quonic`（本地模拟器开箱即用）。

```python
from quonic import qgate, qif, qshow
from quonic.gates import CX, H, I, X
```

## 1. 贝尔态（Bell State）

H 制造叠加，CX 制造纠缠。3 行代码得到量子计算最经典的态。

```python
qgate(H, 0)
qgate(CX, 0, 1)
qshow()
```

输出约 50% 的 `|00>` 和 50% 的 `|11>`——两个量子比特被纠缠在一起。

## 2. GHZ 态（3 量子比特）

再串一个 CX，就得到三体纠缠态。

```python
qgate(H, 0)
qgate(CX, 0, 1)
qgate(CX, 1, 2)
qshow()
```

## 3. 量子叠加 if（`qif`）

用 `if / else` 的直觉写量子逻辑。关键是控制比特**不被测量**，
两个分支相干叠加，产生真纠缠态而非经典混合态。

```python
qgate(H, 0)                        # 控制比特进入叠加态 (|0>+|1>)/√2
qif(0).then(X, 1).else_(I, 1)      # q0==1 翻转 q1，否则不动
qshow()
```

## 4. 量子整数（`QInt`）

把一个整数加载进量子寄存器，支持经典加载、均匀叠加、量子加法。

```python
from quonic import QInt

x = QInt(3, value=5)   # |5> = |101>
x.h()                  # 均匀叠加到 |0>..|7>
x += 3                 # 每个分量加 3（模 8）
qshow()
```

## 5. 一键切换后端

同一段代码，改一个参数跑在不同后端上，输出一致。

```python
qgate(H, 0)
qgate(CX, 0, 1)
qshow(backend="cirq")      # 换成 Google 生态的 Cirq
# qshow(backend="pennylane")  # 或 PennyLane
# qshow(backend="native")     # 或 QuoNic 自研引擎
```

## 6. 算法模板：Grover 搜索

内置算法模板，填空即可运行。先装依赖：`pip install 'quonic[algorithms]'`。

```python
from quonic.algorithms import grover

result = grover("11", 2, shots=1024)   # 在 2 比特中搜索 |11>
result.counts
```
